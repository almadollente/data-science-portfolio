```python
# download, extract, load, and remove artifacts
import os
import shutil
import zipfile
import pandas as pd
import kaggle

# kaggle dataset 
dataset = 'marvingozo/ethereum-uniswap-v3-btceth-june-2026'
zip_filename = 'ethereum-uniswap-v3-btceth-june-2026.zip'

# target file
target_file = 'data/bars_1m_canonical_v2_2026_06.parquet'

# download specified file within dataset
print("downloading raw zip archive...")
kaggle.api.dataset_download_files(dataset, path='.', unzip=False, force=True)

if os.path.exists(zip_filename):
    print(f"Extracting '{target_file}'...")
    # Extract only the specific file we need
    with zipfile.ZipFile(zip_filename, 'r') as z:
        z.extract(target_file, path='.')
        
    # load
    print("loading...")
    df = pd.read_parquet(target_file)

    # delete from hard drive
    os.remove(zip_filename)
    if os.path.exists('data'):
        shutil.rmtree('data')

    # check 
    print("data loaded")
    display(df.head())

    #explore
    print("columns data types")
    df.info()
    
else:
    print("Error: didn't download")   
    
```

## Data Preparation:
**Format and impute**

```python
# to use minute_ms covert integer datatype to datetime
df['datetime'] = pd.to_datetime(df['minute_ms'], unit='ms')
# datetime as index
df.set_index('datetime', inplace=True)

# impute missing values shown in the difference from above Non-Null counts in 'open', 'high', 'low', and 'close' columns, to 'volume_usd', 'buy_volume_usd', and 'sell_volume_usd'
imputed_volumes = ['volume_usd', 'buy_volume_usd', 'sell_volume_usd']
df[imputed_volumes] = df[imputed_volumes].fillna(0)

# drop unused columns
df.drop(columns = ['minute_ms', 'asset_family', 'dex'], inplace=True, errors='ignore')

# display cleaned data
df.info()
display(df.head())
```

## EDA: Market profiling
**Which specific trading pool processes the highest capital to determine the most reliable baseline for real-time markey prices?**
**Are there smaller, less active markets that fall behind and create a temporary price gap that can be used to automate a profit?** 

```python
# group pool_name and fee_tier to calculate total trading volume and observation counts, then sort in descending order
pool_stats = df.groupby(['pool_name', 'fee_tier']).agg(total_observations=('close', 'count'), total_volume_usd=('volume_usd', 'sum')).sort_values('total_volume_usd', ascending=False)

# change the volume to millions (M)
pool_stats['total_volume_usd (Millions)'] = (pool_stats['total_volume_usd'] / 1_000_000).round(2)
pool_stats.drop(columns=['total_volume_usd'], inplace=True)
display(pool_stats)
```

I can infer that large institutional trades will execute in the weth_usdc_uni_v3_500 pool since it has the deepest liquidity (`$2.7 Billion` in volume), compared to weth_usdc_uni_v3_3000 pool trades the same asset pair but with substantially less liquidity(`$129 Million`). This massive disparity causes weth_usdc_uni_v3_3000 pool trades to lag in price discovery during major market moves, and this creates temporary price gaps, which is the profit window this arbitrage will target.

```python
# The focal point

# using the anchor pool (biggest liquidity) and the target pool (the smallest liquidity trading that same asset pair)
weth_usdc_pools = df[df['pool_name'].isin(['weth_usdc_uni_v3_500', 'weth_usdc_uni_v3_3000'])]

# make each pool's closing price a separate column to compare prices by time
price_comparison = weth_usdc_pools.pivot_table(index='datetime', columns='pool_name', values='close')

# calculate the arbitrage gap
price_comparison['price_spread_usd'] = abs(price_comparison['weth_usdc_uni_v3_500'] - price_comparison['weth_usdc_uni_v3_3000'])

# drop when one pool didn't trade
price_comparison.dropna(inplace=True)

# show the top 10  arbitrage windows to see how wide the gap got during highest volatility in June 2026
display(price_comparison.sort_values(by='price_spread_usd', ascending=False).head(10))
```

## Arbitrage anomalies and market shocks visualization 

```python
import matplotlib.ticker as ticker
import matplotlib.pyplot as plt
import pandas as pd

# get daily volume for both pools
vol_500 = df[df['pool_name'] == 'weth_usdc_uni_v3_500']['volume_usd'].resample('D').sum()
vol_3000 = df[df['pool_name'] == 'weth_usdc_uni_v3_3000']['volume_usd'].resample('D').sum()

# daily volume df
daily_volume = pd.DataFrame({'Anchor Pool (500 Fee - Deep Liquidity)': vol_500, 'Target Pool (3000 Fee - Shallow Liquidity)': vol_3000}).fillna(0)

# stacked area chart to show volume dominance
daily_volume.plot.area(alpha=0.3, color=['blue', 'teal'])

plt.title("Daily Capital Flow: Establishing the Market Baseline (June 2026)", fontweight='bold')
plt.xlabel("Date")
plt.ylabel("Daily Trading Volume (USD)")

# format y-axis to show in millions (M)
formatter = ticker.FuncFormatter(lambda x, pos: f'${x*1e-6:.0f}M')
plt.gca().yaxis.set_major_formatter(formatter)
plt.show()

```

The trading pool with the deepest liquidity proves to be the market baseline, whereas the the trading pool with the shallow liquidity shows to not trade at the same efficiency (true market price) during high volume shocks, creating arbitrage windows.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# anchor pool and target pool pairplot 
# chart legend
pair_data = price_comparison.rename(columns={
    'weth_usdc_uni_v3_500': 'Anchor Pool Price ($)', 
    'weth_usdc_uni_v3_3000': 'Target Pool Price ($)', 
    'price_spread_usd': 'Arbitrage Spread ($)'})
# KDE on the diagonal show where prices stayed the longest
pair_plot = sns.pairplot(pair_data, diag_kind='kde', plot_kws={'alpha': 0.4, 's': 15, 'edgecolor': 'none'}, height=3.5)

pair_plot.fig.suptitle("WETH/USDC Price Parity & Arbitrage Spread Distribution", y=1.05, fontweight='bold')

plt.show()
```

The middle graph in the market pool price row, shows when the the pools are synced (market efficiency), and the scattered dots reveal the temporary price gaps. This is confirmed in the bottom row showing spread anomalies spiking up to `$2000`

```python
import seaborn as sns
import matplotlib.pyplot as plt

heatmap_df = price_comparison.copy()
heatmap_df['hour'] = heatmap_df.index.hour
heatmap_df['day'] = heatmap_df.index.day_name()

days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# find outliers from the maximum spread for each hour/day intersection
heatmap_data = heatmap_df.pivot_table(index = 'day', columns='hour', values='price_spread_usd', aggfunc='max').reindex(days_order)

sns.heatmap(
    heatmap_data, 
    cmap='YlOrRd', 
    linewidths=0.5, 
    annot=False, 
    cbar_kws={'label': 'Max Arbitrage Spread (USD)'})

plt.title("Arbitrage Heatmap: Maximum Price Spreads by Hour of Day (June 2026)", fontweight='bold')
plt.xlabel("Hour of Day (0-23)")
plt.ylabel("Day of Week'")
plt.show()
```

Arbitrage anomalies occur most frequently on Mondays, making it the most volatile day of the week. Across the entire week, 02:00 and 22:00 consistently emerge as distinct trigger times for temporary price gaps

```python
import plotly.express as px
import pandas as pd

# interactive box plot
fig = px.box(
    heatmap_df, 
    x='day', y='price_spread_usd',
    category_orders={'day': days_order},
    title='<b>Core Arbitrage Spread Distribution by Day (Capped at $100)</b>', 
    labels={'day': 'Day of Week', 'price_spread_usd': 'Arbitrage Spread (USD)'},
    color='day',
    color_discrete_sequence=px.colors.qualitative.Dark2
)

# hide outliers to only show core distributions 
fig.update_layout(yaxis=dict(range=[-5, 100]), showlegend=False)
fig.show()
```

By focusing the visual weight on the core distribution, the box plot illustrates that the normal resting state of the market is highly efficient with median spreads tightly bounded near zero. However, the presence of outliers proves that price parody breakdowns or distinct temporary structural events, rather than standard market variance (even after extreme $1000+ anomalies are filtered out). This tight baseline efficiency makes these arbitrage windows great target for a ML anomaly detection model.

```python
import matplotlib.pyplot as plt

# 60-minute rolling correlation between the two pool prices
rolling_corr = price_comparison['weth_usdc_uni_v3_500'].rolling(window=60).corr(price_comparison['weth_usdc_uni_v3_3000'])

plt.plot(price_comparison.index, rolling_corr, color='darkgreen')
plt.title("Market Coupling: 60-Minute Rolling Price Correlation", fontweight='bold')
plt.xlabel("Date")
plt.xticks(rotation=45) 
plt.ylabel("Correlation Coefficient")

# highlight correlation
plt.axhline(1.0, color='black')
plt.axhline(0.95, color='red', label="Uncoupling Threshold")
plt.legend(loc='lower right')
plt.show()

```

This shows short term decoupling between the two trading pools. The baseline correlation remains near 1, but the any drop below the 0.95 threshold reveals localized liquidity failures. This uncoupling acts as a highly predictive input featuree for a classification model to predict when to execute a trade.
